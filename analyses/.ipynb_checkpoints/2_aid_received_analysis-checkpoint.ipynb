# Aid Received Analysis

This notebook will explore and analyse the aid data from the aid_events mart, which is based on data from: https://data.humdata.org/dataset/state-of-palestine-gaza-aid-truck-data

This page is currently still a work in progress.

```python
import duckdb
import pandas as pd
import matplotlib.pyplot as plt

con = duckdb.connect("../dev.duckdb")
```

```python
x = con.execute("""
SELECT
    number_of_trucks,
    items,
    cargo_category,
    quantity,
    units,
    donation_type,
    crossing,
    organisation_name,
    organisation_role
FROM fct_aid_received f
JOIN bridge_organisation b
  ON f.aid_event_id = b.aid_event_id
JOIN dim_organisation o
  ON b.org_id = o.org_id
WHERE organisation_name = 'JICA'
""").df()

print(x)
```

```python
df = con.execute("""
SELECT
    year,
    week_of_year,
    cargo_category,
    SUM(quantity) AS total_pallets
FROM main_mart.aid_events
WHERE units = 'Pallets'
GROUP BY year, week_of_year, cargo_category
ORDER BY year, week_of_year
""").df()

print(df)



df["week_start"] = pd.to_datetime(
    df["year"].astype(str) + "-W" + df["week_of_year"].astype(str) + "-1",
    format="%G-W%V-%u"
)


df_pivot = df.pivot(
    index="week_start",
    columns="cargo_category",
    values="total_pallets"
).fillna(0)


plt.figure(figsize=(12,6))

for column in df_pivot.columns:
    plt.plot(df_pivot.index, df_pivot[column], label=column)

plt.xlabel("Date")
plt.ylabel("Total Pallets")
plt.title("Aid Received Over Time (Pallets) by Cargo Category")
plt.legend()
plt.tight_layout()
plt.show()
```
